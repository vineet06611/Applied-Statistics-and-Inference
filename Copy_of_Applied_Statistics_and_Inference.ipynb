impact decision-making ?

Type I and Type II errors are false conclusions in hypothesis testing. A Type I error (false positive,
) occurs when a true null hypothesis is incorrectly rejected, leading to false discoveries (e.g., claiming a drug works when it doesn't). A Type II error (false negative,
) occurs when a false null hypothesis is not rejected, missing true effects     (e.g., missing a disease).

    Key Concepts in Hypothesis Testing
    Null Hypothesis : The default position, typically suggesting no effect or no difference.
    Type I Error (False Positive): You reject
    when it is actually true.

    - Significance Level : The probability of making a Type I error (commonly set at 0.05 or 5%).

    Type II Error (False Negative): You fail to reject when it is actually false.

    - Power of the Test : The probability of correctly rejecting a false null hypothesis.

Impact on Decision-Making

 -Type I Error Consequences: Leads to misguided actions, such as investing in a useless project, releasing a faulty product, or wrongly convicting an innocent person.

 -Type II Error Consequences: Results in missed opportunities, such as ignoring a effective treatment, dismissing a valid business strategy, or failing to identify a real problem.

 -The Trade-off: Reducing one type of error increases the other. A more conservative test (lower
) reduces Type I errors but increases Type II errors, and vice-versa

4. What is a confidence interval, and how does the margin of error influence
its width and interpretation ?

A confidence interval (CI) is a range of values that likely contains a population parameter (like an average or a percentage) based on a sample from that population. It’s essentially saying, "I don't know the exact number, but I'm 95% sure it falls between X and Y."
The margin of error (MOE) is the "plus or minus" part of that calculation. It directly dictates the width and precision of the interval:
Width: The MOE is exactly half the width of the confidence interval. If your MOE increases, the interval gets wider and less specific. If the MOE decreases, the interval narrows.
Interpretation: A smaller MOE means you have a more precise estimate. For example, saying a candidate has



 support (interval: 48–52%) is much more informative than saying they have



 support (interval: 40–60%).
Trade-off: To get a smaller MOE (and a narrower, better interval), you generally need a larger sample size or you must be willing to accept a lower confidence level (e.g., being only 90% sure instead of 95%).

5. Describe the purpose and assumptions of an ANOVA test. How does it
extend hypothesis testing to more than two groups?

The primary purpose of an ANOVA (Analysis of Variance) test is to determine whether there are statistically significant differences between the means of three or more independent groups. While a t-test is limited to comparing exactly two groups, ANOVA allows researchers to analyze multiple groups simultaneously to see if at least one group mean is different from the others.

Key Assumptions

For ANOVA results to be valid, the following assumptions must generally be met:

Independence of Observations: The data points in each group must be independent of one another and the groups themselves must be unrelated (no overlap in participants).

Normality: The dependent variable should be approximately normally distributed for each group. This is typically visualized with Q-Q plots or tested with a Shapiro-Wilk test.

Homogeneity of Variance (Homoscedasticity): The variance (spread) of the data should be roughly equal across all groups. This is often verified using Levene’s test or Bartlett’s test.

Continuous Dependent Variable: The outcome being measured must be continuous (interval or ratio level), such as height, weight, or test scores.

Extending Hypothesis Testing to 3+ Groups

ANOVA extends hypothesis testing by shifting the focus from comparing individual means to analyzing variability. It uses the F-statistic, which is a ratio of two types of variance:

Between-Group Variance: How much the means of the various groups differ from the overall "grand mean".

Within-Group Variance: The natural spread or "noise" of data points within each specific group.

If the variance between the groups is significantly larger than the variance within the groups, the F-statistic will be high, leading to the rejection of the null hypothesis.

Why not just use multiple t-tests?

Performing separate t-tests for every possible pair of groups (e.g., A vs. B, B vs. C, and A vs. C) leads to Alpha Inflation. Every time you run a test at a 5% significance level, there is a 5% chance of a Type I error (a false positive). Running many tests increases this cumulative risk dramatically. ANOVA acts as an "omnibus" test that maintains a single, controlled error rate for the entire set of comparisons.

6. Write a Python program to perform a one-sample Z-test and interpret the
result for a given dataset?

To perform a one-sample Z-test in Python, you can calculate the Z-statistic and p-value using libraries like numpy and scipy.stats. A Z-test is used when the population standard deviation is known and the sample size is typically greater than 30.

Python Code for One-Sample Z-Test

The following program calculates the Z-score and p-value for a given dataset compared to a population mean.

    python

    import numpy as np
    
    from scipy import stats

def perform_z_test(data, pop_mean, pop_std, alpha=0.05):
    """
    Performs a one-sample Z-test.
    """
    n = len(data)
    sample_mean = np.mean(data)
    
    # Formula: z = (x_bar - mu) / (sigma / sqrt(n))
    z_score = (sample_mean - pop_mean) / (pop_std / np.sqrt(n))
    
    # Calculate two-tailed p-value
    p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
    
    # Interpretation
    print(f"Sample Mean: {sample_mean:.2f}")
    print(f"Z-Statistic: {z_score:.4f}")
    print(f"P-Value: {p_value:.4f}")
    
    if p_value < alpha:
        print(f"Result: Reject the null hypothesis (Significant difference at alpha={alpha}).")
    else:
        print(f"Result: Fail to reject the null hypothesis (No significant difference).")

# Example Usage
my_data = [28, 32, 24, 26, 29, 27, 30, 25, 24, 35] # Your dataset
population_mean = 25
population_standard_deviation = 3.0

perform_z_test(my_data, population_mean, population_standard_deviation)

9. Write a Python function to calculate and visualize the confidence interval
for a sample mean?

You can calculate a confidence interval for a sample mean using scipy.stats.t.interval and visualize it using matplotlib's axvline or errorbar functions. The t-distribution is used because the population standard deviation is typically unknown.

import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def calc_and_visualize_ci(data, confidence_level=0.95):
    """
    Calculates the confidence interval for a sample mean and visualizes it.

    Args:
        data (array-like): The sample data.
        confidence_level (float): The desired confidence level (e.g., 0.95).

    Returns:
        tuple: (sample_mean, ci_lower, ci_upper)
    """
    sample_mean = np.mean(data)
    n = len(data)
    # Use Bessel's correction (ddof=1) for sample standard deviation
    sample_std = np.std(data, ddof=1)
    # Calculate the standard error of the mean (SEM)
    standard_error = sample_std / np.sqrt(n)
    
    # Calculate the confidence interval using the t-distribution
    # loc is the sample mean, scale is the standard error, df is degrees of freedom
    ci_lower, ci_upper = stats.t.interval(
        alpha=confidence_level,
        df=n - 1,
        loc=sample_mean,
        scale=standard_error
    )
    
    print(f"Sample Mean: {sample_mean:.2f}")
    print(f"{confidence_level*100:.0f}% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
    
    # Visualization
    plt.figure(figsize=(8, 5))
    plt.hist(data, bins=10, color='skyblue', alpha=0.7, label='Sample Data Distribution')
    
    # Plot the sample mean
    plt.axvline(sample_mean, color='green', linestyle='solid', linewidth=2, label='Sample Mean')
    
    # Plot the confidence interval boundaries
    plt.axvline(ci_lower, color='red', linestyle='dashed', linewidth=2, label='Confidence Interval Bounds')
    plt.axvline(ci_upper, color='red', linestyle='dashed', linewidth=2)
    
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.title(f'Sample Mean and {confidence_level*100:.0f}% Confidence Interval')
    plt.legend()
    plt.show()

    return sample_mean, ci_lower, ci_upper

# --- Example Usage ---
# Generate sample data (e.g., exam scores from a normal distribution)
np.random.seed(42)
exam_scores = np.random.normal(loc=80, scale=10, size=50)

# Calculate and visualize the 95% confidence interval
mean, lower_bound, upper_bound = calc_and_visualize_ci(exam_scores, confidence_level=0.95)


10. Perform a Chi-square goodness-of-fit test using Python to compare
observed and expected distributions, and explain the outcome?

To perform a Chi-square goodness-of-fit test in Python, you can use the scipy.stats.chisquare function with your observed and expected frequencies. The outcome is interpreted by comparing the returned p-value to a pre-determined significance level (alpha, typically 0.05) to decide whether the observed data fits the expected distribution.


from scipy.stats import chisquare
import numpy as np



# Observed frequencies (e.g., actual counts of different colored candies in a bag)
observed_frequencies = np.array([220, 180, 200, 150, 250])

# Expected frequencies (e.g., if we expect equal distribution among 5 colors for 1000 total candies)
total_count = np.sum(observed_frequencies)
num_categories = len(observed_frequencies)
expected_frequency_per_category = total_count / num_categories
expected_frequencies = np.array([expected_frequency_per_category] * num_categories)

print(f"Observed frequencies: {observed_frequencies}")
print(f"Expected frequencies: {expected_frequencies}")



Finally, perform the Chi-square test using the chisquare function from the scipy.stats module:


chi2_statistic, p_value = chisquare(f_obs=observed_frequencies, f_exp=expected_frequencies)

print(f"\nChi-square statistic: {chi2_statistic}")
print(f"P-value: {p_value}")

# Set significance level (alpha)
alpha = 0.05

# Interpret the results
if p_value < alpha:

    print(f"\nResult: Reject the null hypothesis (p < {alpha}).")
    print("Conclusion: The observed distribution is significantly different from the expected distribution.")
else:

    print(f"\nResult: Fail to reject the null hypothesis (p >= {alpha}).")
    print("Conclusion: The observed distribution does not significantly differ from the expected distribution (the fit is 'good enough').")
